## Teoría de conjuntos aplicado a dos equipos de futbol

### Parte 1. Construcción de los conjuntos

1. Lee automáticamente los 20 archivos correspondientes al **Club A** y combínalos en un único conjunto de jugadores.
2. Elimina los jugadores duplicados para que cada futbolista aparezca una única vez.
3. Repite el proceso con los 20 archivos correspondientes al **Club B**.
4. ¿Cuántos jugadores únicos han pertenecido a cada club durante las últimas 20 temporadas?
5. Muestra los conjuntos obtenidos ordenados alfabéticamente.

```python
# Modulo de python para trabajar con el sistema operativo,leer dcumentos...
# se usa para leer carpeas, archivos...
import os

# libreria especializada en el analisis y manejo de datos
import pandas as pd

# Para representarlo con un diagrama de Venn en Python,
# lo más directo es utilizar matplotlib junto con matplotlib-venn.
#import matplotlib.pyplot as plt
#from matplotlib_venn import venn2

# LEER LOS DATOS Y CREAR DOS CONJUNTOS SIN DUPLICADOS

# Lee todos los CSV de una carpeta
# devueLve un conjunto de jugadores únicos.

# Se crea una funcion donde el parametro es la ruta de los archivos a analizar


def carga_datos(path):

    # se crea unconjunto vacio, no lista, no duplicados
    jugadores = set()

    # os.listdir se obtienen los archivos de la carpeta
    # se recorren con un bucl for
    for archivo in os.listdir(path):

        # se comprueba que sean csv, si es devuelve true
        if archivo.endswith(".csv"):

            # Aqui se construye la ruta completa a cada archivo
            ruta = os.path.join(path, archivo)

            # con pandas se abre el csv y se crea un dataframe
            df = pd.read_csv(ruta)

            # dentro del archivo selecciona la columna jugador
            # elimina los valores vacios
            # los convierte a un string
            # los añade al conjunto
            jugadores.update(df["jugador"].dropna().astype(str))

    return jugadores


jugadores_atletico = carga_datos("data/atletico")
jugadores_barcelona = carga_datos("data/barcelona")

# CUANTOS JUGADORES HA HABIDO EN CADA CLUB

print("Jugadores Atletico:", len(jugadores_atletico))
print("Jugadores Barcelons", len(jugadores_barcelona))


# SE MUESTRAN LA LISTA DE JUGADORES ORDENADA ALFABETICAMENTE

print("\nListado Atletico (ordenado):")
print(sorted(jugadores_atletico))

print("\nListado Barcelons (ordenado):")
print(sorted(jugadores_barcelona))
```

### Operaciones con conjuntos

1. Calcula la **unión** de ambos conjuntos.
    - ¿Cuántos jugadores diferentes han pasado por alguno de los dos clubes?
2. Calcula la **intersección**.
    - ¿Qué jugadores han pertenecido a ambos clubes?
    - ¿Cuántos son?
3. Calcula la **diferencia** entre el Club A y el Club B.
    - ¿Qué jugadores han jugado únicamente en el Club A?
4. Calcula la **diferencia simétrica**.
    - ¿Qué jugadores han jugado exclusivamente en uno de los dos clubes?
5. Considerando como **universo** la unión de ambos conjuntos, calcula el **complemento** de la intersección.
    - ¿Qué jugadores no han pasado por ambos clubes?

```python

# Intersección
# Elementos que estan en a y b

interseccion = jugadores_atletico & jugadores_barcelona
print("\nINTERSECCIÓN:", len(interseccion))
print(sorted(interseccion))

# Diferencia jugadores_atletico - jugadores_barcelona
# Seran los elemntos que estan en el atletico no en el barcelona
# aqui importa el orden si no serian los del barcelona y no los del atletico
solo_atletico = jugadores_atletico - jugadores_barcelona
print("\nSOLO Atletico:", len(solo_atletico))

# Diferencia simétrica
# los que solo estan en uno de los dos conjuntos pero no en ambos
diferencia_simetrica = jugadores_atletico ^ jugadores_barcelona
print("\nDIFERENCIA SIMÉTRICA:", len(diferencia_simetrica))

# Unión
# todos los elementos juntos pero sin duplicados
union = jugadores_atletico | jugadores_barcelona
print("\nUNIÓN:", len(union))
```

### Análisis de resultados

1. ¿Qué porcentaje de jugadores del universo ha jugado en ambos clubes?
2. ¿Qué club ha tenido un mayor número de jugadores exclusivos?

```python
# porcentaje que ha jugado en ambos clubes
porcentaje_ambos = (len(interseccion) / len(union)) * 100
print(f"\n% jugadores en ambos clubes: {porcentaje_ambos:.2f}%")

# club con más jugadores exclusivos
# ya teniamos los jugadores que solo estaban en el atletico
# ahora calculamos los del barcelona
solo_barcelona = jugadores_barcelona - jugadores_atletico

# Aqui se compruba cual tiene mas jugadores unicos el taaño de los conjuntos
if len(solo_atletico) > len(solo_barcelona):

    print("\nAtletico tiene más jugadores exclusivos")

elif len(solo_barcelona) > len(solo_atletico):

    print("\nBarcelona tiene más jugadores exclusivos")

else:

    print("\nAmbos clubes tienen el mismo número de jugadores exclusivos")
```
