# ORIGEN DE LOS DATOS SEGUN SU ESTRUCTURA


**1. Tipos de datos:**
- Estructurados, cambian pero con persidad de informacion, fijos. (tablas)
- Semi estructurados, tienen una estructura pero pueden cambiar, son flexibles, se saca lo que se necesita, no es necesario traer todo. (json)
- No estructurados, sin estructura pueden cambiar.
  
**2. Tipos de origen de los datos:**

  Se pueden distinguir viendo quien es el que gestiona la informacion.
  - Fuentes internas
  - Fuentes externas

## EJEMPLO

**Tres Departamentos que recogen datos en diferente formato (csv, xlxs,json) 3 formas de obtener datos pero un uinico objetivo.**


### Unificar los datos

Hay que unificar el formato de los datos, importante a la hora de extraccion, procesamiento y carga.
Ver que campos son coincidentes y estandarixzarlos para igualar la informacion. por ejemplo el campo del id del cliente, si es numerico o no, el formato de las fechas, que hacer con un campo por ejemplo anomalias si es una informacion relevante si mantenerlo o como modificarlo.


### Análisis

1. proceden los datos de la misma fuente
2. se utiliza lo mismo para la identificacion
3. ...

### Integración
Unificar y corregir el sentido de los datos, relacionarlos por campos...
Hay que identificar que campo relacioona los datos para poder unirlos,  puede ser con una clave,
o en nuestro ejemplo un identificador que relaciona un cliente

**Diferenciar Agragacion de Integracion**

Agregacion es juntar sin relacionar. ej copiar los archivos en una misma carpeta.ç, pero los datos siguen separados

**¿Que pasa si hay casos repetidos?**

Depende hay que trabajar con ello y pensar que se va a hacer

Despues ya se puede hacer la trasnformacion cuando la informacion sea lo mas perfecta posible

### Proceso para la Integración
Buscar los datos relacionados por el campo identificador y relacionarlos.
Es recopilar, combinar, transformar y unificar datos para crear un conçjunto unico consistente y preparado para su explotacion.

### Posibles anomalías que encontrar a la hora de unificar datos
1. Formatos
    1. Formatos de fechas
    2. Formatos de cantidades
    3. Formato en el uso de coma o punto
2. Diferentes estructura ( distintos nombres de campos de las fuentes de datos)
3. Duplicados
4. Inconsistencias (uso de mayúscula por ejemplo)
5. Valores nulos
