```python
from itertools import product

# Listas de prendas disponibles
camisetas = ["C1", "C2", "C3"]
pantalones = ["P1", "P2"]
zapatos = ["Z1", "Z2"]

# Generamos todas las combinaciones posibles de outfit
outfits = list(product(camisetas, pantalones, zapatos))

# Mostramos el resultado
for outfit in outfits:
    print(outfit)

print("Número total de outfits:", len(outfits))

# Añadimos una nueva camiseta
camisetas.append("C4")

# Recalculamos combinaciones
outfits_actualizados = list(product(camisetas, pantalones, zapatos))

print("Nuevo número de outfits:", len(outfits_actualizados))
```
