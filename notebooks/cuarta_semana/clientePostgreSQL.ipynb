## Conceptos previos

Docker. Permite ejecutar aplicaciones aisladas en contenedores. En esta práctica tendremos un contenedor para PostgreSQL y otro para pgAdmin.

PostgreSQL. Es el servidor que almacena y gestiona las bases de datos, las tablas y sus regist

Vamos a crear dos contenedores uno para cada cosa

## DOCKER-COMPOSE
Necesitamos crear el fichero docker-compose.yml
Es un archivo de configuración que describe los contenedores que queremos ejecutar, sus puertos, las contraseñas de acceso y la comunicación entre ellos.

En lugar de configurar cada contenedor manualmente, Docker Compose leerá este archivo y levantará todo el entorn
En el docker-compose definimos los servicios con los que vamos a trabajar en el contenedor, aqui vamos a trabajar con el servidor de base de datos y con postgresql definimos la imagen que descargar al contenedor, las variables de entorno...

### Arrancar los contenedores

Desde la terminal, situado en la carpeta donde has guardado el archivo, ejecuta:

```
docker compose up -d
```

Docker descargará las imágenes si es necesario y creará los dos contenedores.

El parámetro `-d` significa que se ejecutarán en segundo plano.

Comprueba que están funcionando:

```
docker
Deberías encontrar los servicios `postgres` y `pgadmin` con estado de ejecución. La primera vez, la descarga de las imágenes puede tardar unos minutos.
```
