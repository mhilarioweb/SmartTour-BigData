## Modelando relaciones reales Relación Jugador → Club → Temporada

A partir de los ficheros de jugadores:

### Construye la relación completa:

```python
# importa os
import os

# importa pandas
import pandas as pd

# modulo de python con estructuras de datos adicionales
# se importa defaultdic es como un diccionario
from collections import defaultdict

# libreria para trabajar con grafos y redes
import networkx as nx

# para crear grafos y visualizaciones
import matplotlib.pyplot as plt

# para construit la relacion usamos un diccionario anidado
# relación: jugador -> lista de (club, temporada)
relacion = defaultdict(list)


# la relacion sera entre cada jugador y a temporada a la que pertenece
def cargar_relacion(ruta, club):

    # de nuevo se buscan los archivos csv en la carpeta
    for archivo in os.listdir(ruta):
        if archivo.endswith(".csv"):

            # Se coje el nombre de la temporada del nombre del fichero
            # se sustituye el .csv po "" y asi quedara por ejemlpo 2023-2024
            temporada = archivo.replace(".csv", "")

            # se leen los csv de la ruta
            df = pd.read_csv(os.path.join(ruta, archivo))

            # se recorre el conjunto con un for borrando los que estan en blanco
            for jugador in df["jugador"].dropna():

                # se va introduciendo en el diccionario y cada uno tiene una lista asociada con
                # la temporada y club en que estaban
                relacion[jugador].append((club, temporada))


# SE CARGAN LOS JUGADORES
cargar_relacion("data/atletico", "Atletico")
cargar_relacion("data/barcelona", "Barcelona")

```

### Representa esa relación de tres formas:
    - tabla
    - diccionario (lista en Python o R)
    - grafo simple

1. COMO UNA TABLA

```python
# COMO UNA TABLA

# se crea lo primero una lista vacia
tabla = []

# Coge el diccionario de antes ylo convierte en una tabla de pandas
# con u n for se recorre el diccionario
# con items se puede recorrer el diccionario obteniendo clave - valor
# teniamos los jugadores asiciados a listas en las que estaba el equipo y la temporada
for jugador, valores in relacion.items():

    # recorre la lista a la que estaba asociada a cada jugador
    # asi y apodemos obtener el jugador la temporada y el lcub
    for club, temporada in valores:
        tabla.append([jugador, club, temporada])

# lo convertimos en un dataframe
df_relacion = pd.DataFrame(tabla, columns=["Jugador", "Club", "Temporada"])

print(df_relacion.head())

```

2. COMO UN DICCIONARIO

```python

# COMO UN DICCIONARIO

# convertimos eldefaultdict en un diccionario normal de python
print(dict(relacion))
```

3. COMO UN GRAFO SIMPLE

```python


# COMO UN GRAFO SIMPLE

# se crea el grafo
G = nx.Graph()

# se recorre la relacion con un for y con items
for jugador, valores in relacion.items():

    # se recorre la lista que habia quedado asociada al jugador en el diccionario anidado
    for club, temporada in valores:

        # se añaden los elementos como nodos y se les pone el atributo segun corresponda
        G.add_node(jugador, type="jugador")
        G.add_node(club, type="club")
        G.add_node(temporada, type="temporada")

        # se crean las relaiones del jugador con el club y la temporada
        G.add_edge(jugador, club)
        G.add_edge(jugador, temporada)


# se crea la figura, con figsize se establece el tamaño
plt.figure(figsize=(10, 6))

# se dibuja el grafo, la bels escribe el nombre d ecada nodo y size el tamño visual de los nodos
nx.draw(G, with_labels=True, node_size=1500)

# se muestra el grafo por pantalla
plt.show()
```

1. ¿QUË JUGADOR TIENE MAS TEPORADAS?

```python
#importamos counter de modulo collections de python
#Es una estructura parecida a un diccionario diseñada principalmente para realizar conteos
from collections import Counter

#se crea un contador vacio
contador_temporadas = Counter()

#con un for se recorren los jugadores con items para recoger el valor y la clave
for jugador, valores in relacion.items():
    
    #Aquí están ocurriendo dos operaciones.
    #v va a ser recorrer las tuplas y v[1] el segundo elemento de cada una
    #asi en temporadas creamos un conjunto en el que se eliminan los duplicados
    temporadas = set([v[1] for v in valores])

    #a cada jugador se le van añadiendo las temporadas
    contador_temporadas[jugador] = len(temporadas)

# máximo
max_temporadas = max(contador_temporadas.values())

# ya se calculan los máximos
jugadores_top = [j for j, v in contador_temporadas.items() if v == max_temporadas]

print("Jugadores con más temporadas:", jugadores_top)
print("Número de temporadas:", max_temporadas)
```
