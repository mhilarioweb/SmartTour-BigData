# Análisis básico de clientes integrando múltiples fuentes de datos

## Ejercicio

Tenemos tres ficheros con datos en diferentes estructuras de fuentes, tenemos que hacer la integracion de los datos.

Un csv con cliente_id y producto_id, un excel con producto_id y un json con cliente_id

### Pasos que tenemos que seguir:

Necesitamos hacer un script en python que nos permita traer toda la informacion que tenemos en una sola fuente de informacion, es una integraciond de datos:

1. En primer lugar tenemos que leer los ficheros con el script

**En primer lugar vamos a leer los ficheros**

1. Se lee con la **librería pandas**, lo primero es importarla.
2. Después hay que cargar las fuentes de datos:
    - Dentro de la libreria **pandas existen funciones para leer** los distintos tipos de archivos
    - Los **ficheros que se cargan se convierten** en una estructura para que se pueda leer
    - Para visualizar la informacion podemos usar **print o display**

```python
#En primer lugar se importa pandas
import pandas as pd

#Funciones de pandas para leer ficheros:

# PARA LEER UN CSV
ventas = pd.read_csv("datos/ventas.csv")

# PARA LEER UN EXCEL
productos = pd.read_excel("datos/productos.xlsx")

# PARA LEER UN JSON
opiniones = pd.read_json("datos/opiniones.json")

# PARA VISUALIZARLOS

#display(ventas)
#print(productos)
#print(opiniones)
```

**En segundo lugar un control de calidad de los datos que tenemos cargados**

- Vamos a comprobar si hay valores nulos o si existen duplicados.
- Saber si el dataframe tiene valores correctos.

**1. Informacion sobre los datos**

Podemos trabajar los dataframes con la **funcion info()**: nos da el tipo de datos y la informacion de los datos que tengo en cada columna


```python
#Control de calidad de los datos

print ("-------------------")
print("DATAFRAMES EN VENTAS")
print ("-------------------") 

print(ventas.info())

print ("-------------------")
print("DATAFRAMES EN PRODUCTOS")
print ("-------------------") 

print(productos.info())

print ("-------------------")
print("DATAFRAMES EN OPINIONES")
print ("-------------------") 

print(opiniones.info())
```

**2. Comprobar si hay valor nulos**

**isnull()** pregunta si es nulo y la suma de todos los nulos nos da si hay o no valores nulos

```python
#Comprobar valores nulos

print ("-------------------")
print("VALORES NULOS EN VENTAS")
print ("-------------------")

print(ventas.isnull().sum())

print ("-------------------")
print("VALORES NULOS EN PRODUCTOS")
print ("-------------------")

print(productos.isnull().sum())

print ("-------------------")
print("VALORES NULOS EN OPINIONES")
print ("-------------------")

print(opiniones.isnull().sum())
```

**3. Comprobar valores duplicados**

**duplicated()** es para ver si hay duplicados y sumandolos con **sum()** para ver cuantos hay en cada dataframe

```python
#Comprobar valores duplicados

print ("-------------------")
print("VALORES DUPLICADOS EN VENTAS")
print ("-------------------")

print(ventas.duplicated().sum())

print ("-------------------")
print("VALORES DUPLICADOS EN PRODUCTOS")
print ("-------------------")

print(productos.duplicated().sum())

print ("-------------------")
print("VALORES DUPLICADOS EN OPINIONES")
print ("-------------------")

print(opiniones.duplicated().sum())
```

**En tercer lugar unificamos los datos**

Se pueden convinar varios dataframe, integrarlos seria enlaazrlos lo hacemos con la funcion **merge()** que une dos estructuras de datos y las une a traves de un llave.
Seria un dataframe otro dataframe y la clave, columna o valor por el que las voy a unir, si tengo dos claves la puedo unir por dos, si el nombre de la clave es distinto hay que especificar el nombre de las dos claves.
En SQL en el **join** left join o right join (ventas a productos o productos a ventas).

**1. La primera unificacion seria ventas y productos**

Ahora lo que tendremos es un merge de ventas y productos por producto_id para que los productos se añadan a las ventas , se unen utilizando ventas como la tabla principal.

```python
#Unificacion datps coon ventas como tabla principal

ventas_productos = pd.merge(ventas, productos, on="producto_id")

#mostramos ventas productos la unificacion de ventas con productos

print("---------------------------------")
print("Unificacion de ventas y productos")
print("---------------------------------")

display(ventas_productos)
```

**2. La segunda unificacion seria ventas y productos**

Para completad la unificacion tenemos que unir las opiniones de los clientes que han realizado las compras, pero ahora trabajamos con ventas productos a las que le añadimos las opiniones, sigue siendo un left porque la tabla principal sigue siendo ventas.
El campo que las une es **cliente_id**

```python
#Unificacion datps coon opiniones, ventas sigue siendo la tabla principal

datos_completos = pd.merge(ventas_productos, opiniones, on="cliente_id")

#mostramos ventas productos la unificacion de ventas con productos

print("-------------------------")
print("Unificacion con opiniones")
print("-------------------------")

display(datos_completos)

```

**Por ultimo vamos a completar la informacion**

Vamos a añadir una columna que sea el total por ejemplo.
Se pueden añadir campos al dataframe para luego trabajadçr con el de formamas eficiente

```python
## Total Gastado por cliente
datos_completos["total"] = datos_completos["cantidad"] * datos_completos["precio"]
gasto_cliente = datos_completos.groupby("cliente_id")["total"].sum()
print(gasto_cliente)
```

**Podriamos ya realizar un analisis de la información**

```python
#Valoración media por cliente
#se agrupa la valoracion por id de cliente y se hace la suma de las valoraciones
valoracion_media = datos_completos.groupby("cliente_id")["valoracion"].mean()
print(valoracion_media)

#Producto más vendido
#Se cuenta el nombre que mas se repite
producto_popular = datos_completos["nombre"].value_counts()
print(producto_popular)
```
