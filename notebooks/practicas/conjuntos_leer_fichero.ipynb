## Contrucción de conjuntoss

Creamos una funcion para la carga de los alumnos desde los ficheros
Se crea un conjunto con set() ya que cuando carguemos todos los datos no podrá haber duplicados
Con el modulo de os python puede interactuar con el SO del equipo, lo usamos para buscar los ficheros en los directorios
De la libreria de pandas usamos metodo para leer el csv
Con el modulo csv trabajamos con csv

```python
import os
import pandas as pd

import csv

grupo_a = set()
grupo_b = set()

#Lo primero abrir el archivo
with open("grupo_a.csv","r") as archivo:

    #lo leemos
    lector =csv.reader(archivo)

    #vamos recorriendo las filas y lo añadimos al conjunto
    for fila in lector:
        grupo_a.add(fila[0])
#imprimimos el conjunto
print(grupo_a)

#Para el conjunto b lo mismo:

with open("grupo_b.csv","r") as archivo:
    lector =csv.reader(archivo)
    for fila in lector:
        grupo_b.add(fila[0])

print(grupo_b)
```

1. Lee automáticamente los 20 archivos correspondientes al **Grupo A** y combínalos en un único.
2. Elimina los duplicados para que cada futbolista aparezca una única vez.
3. Repite el proceso con los 20 archivos correspondientes al **Grupo B**.
4. ¿Cuántos únicos han pertenecido a cada club durante las últimas 20 temporadas?
5. Muestra los conjuntos obtenidos ordenados alfabéticamente.

```python
# Para ver cuantos unicos pertenecen al grupo se cuenta la longitud de los
#conjuntos ya que en un conjunot no hay repetidos
print("únicos grupo A:", len(grupo_a))
print("Jugadores únicos Club B:", len(grupo_b))

# Para mostrarlos ordenados se usa el metodo sorted
print("\nListado Club A (ordenado):")
print(sorted(grupo_a))

print("\nListado Club B (ordenado):")
print(sorted(grupo_b))
```

## Operaciones con conjuntos

1. Calcula la **unión** de ambos conjuntos.
    - ¿Cuántos jugadores diferentes han pasado por alguno de los dos clubes?
2. Calcula la **intersección**.
    - ¿Qué jugadores han pertenecido a ambos clubes?
    - ¿Cuántos son?
3. Calcula la **diferencia** entre el Club A y el Club B.
    - ¿Qué jugadores han jugado únicamente en el Club A?
4. Calcula la **diferencia simétrica**.
    - ¿Qué jugadores han jugado exclusivamente en uno de los dos clubes?
5. Considerando como **universo** la unión de ambos conjuntos, calcula el **complemento** de la intersección.
    - ¿Qué jugadores no han pasado por ambos clubes?

```python
#Union de los conjuntos
union_grupos = grupo_a | grupo_b

print(union_grupos)

#Interseccion de los conjuntos
interseccion_grupos = grupo_a & grupo_b

print(interseccion_grupos)

#Diferencia de los conjuntos
diferencia_grupos = grupo_a | grupo_b

print(diferencia_grupos)

#Diferencia simetrica de los conjuntos
diferencia_simetrica_grupos = grupo_a ^ grupo_b

print(diferencia_simetrica_grupos)

union_grupos = grupo_a | grupo_b
interseccion_grupos = grupo_a & grupo_b

#Complemento de los conjuntos

complemento_interseccion = union_grupos - interseccion_grupos
print("\nCOMPLEMENTO INTERSECCIÓN:", len(complemento_interseccion))

```
