# Practica de conjuntos

Creamos dos conjuntos, se crean con la funcion set()

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);
```

## Vamos a calcular la union
Para calcular la union de dos conjuntos se usa el operador |
Es la combinacion de los elementos unicos de a y b, osea se combinan sin duplicados.
Cantidad total sin contarlos dos veces

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);

union_grupos = grupo_a | grupo_b

print(union_grupos)
```

## Vamos a calcular la interseccion
Para ello se usa el operador &
Extra los elementos presentes simultaneamente en ambos conjuntos

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);

interseccion_grupos = grupo_a & grupo_b

print(interseccion_grupos)
```

## Vamos a calcular la diferencia
Para ello se usa el operador -
Los que estan solo en un u otro, el rden importa se vera los del primero

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);

diferencia_grupos = grupo_a - grupo_b

print(diferencia_grupos)
```

## Vamos a calcular la diferencia simetrica
Para ello se usa el operador ^
Estan en a o en b pero no en ambos

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);

diferencia_grupos = grupo_a ^ grupo_b

print(diferencia_grupos)
```

## Vamos a calcular el complemento de la interseccion respecto al universo
Se usara la union y la interseccion y Para ello se usa el operador -

```python
grupo_a = set(["Maria", "Angel","Pablo","Laura","Marta"]);
grupo_b = set(["Patricia", "Angel","Cristina","Laura","Martin"]);

union_grupos = grupo_a | grupo_b
interseccion_grupos = grupo_a & grupo_b

print("UNION: " , union_grupos)
print("INTERSECCION: " , interseccion_grupos)

complemento_interseccion = union_grupos - interseccion_grupos
print("\nCOMPLEMENTO INTERSECCIÓN:", len(complemento_interseccion))
```
