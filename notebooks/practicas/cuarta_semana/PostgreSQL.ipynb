## Matriz de filtrado
Tengo un grupo de registros pero no quiero trabajar con todos solo con un subconjunto tenemos tres clausulas, **WHERE** siempre antes de agrupar, **HAVING** despues de agrupar, 
y **FILTER**

```python
# WHERE
SELECT CITY, 
    AVG(TEMP) AS TEMPERATURA_MEDIA
    FROM WEATHER 
    WHERE TEMP > 15 
    GROUP BY CITY;

# HAVING
SELECT CITY, 
    AVG(TEMP) AS TEMPERATURA_MEDIA
    FROM WEATHER 
    GROUP BY CITY
    HAVING AVG(TEMP) > 15; 

# FILTER
SELECT CITY, 
    AVG(TEMP) AS MEDIA_GENERAL
    AVG(TEMP) FILTER(
        WHERE TEMP > 20
    )AS MEDIA_SUPERIOR_20
    FROM WEATHER 
    GROUP BY CITY;
```
