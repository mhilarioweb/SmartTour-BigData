# Elección de outfit - combinatoria en Big Data

Una tienda de moda online quiere analizar todas las posibles combinaciones de ropa que puede ofrecer a sus clientes para generar recomendaciones automáticas de outfits. Para simplificar el problema, se dispone de los siguientes elementos:

- 3 camisetas: C1, C2, C3
- 2 pantalones: P1, P2
- 2 pares de zapatos: Z1, Z2

A partir de este conjunto de prendas, el sistema debe analizar todas las posibles combinaciones de outfits completos formados por una camiseta, un pantalón y un par de zapatos.

### Principio multiplicativo

1. Calcula cuántos outfits diferentes se pueden formar en total utilizando el principio multiplicativo.
2. Explica paso a paso cómo se obtiene el resultado a partir del número de opciones en cada categoría de ropa.
3. Interpreta el resultado en el contexto de un sistema de recomendación de moda online.

```python
# modulo de pyton para trabajar con herramientas de iteraciones y combinaciones
# de este modulo importamos solo poducts
from itertools import product

# importamos pandas
import pandas as pd

# Definimos los datos
# Listas de prendas disponibles
camisetas = ["C1", "C2", "C3"]
pantalones = ["P1", "P2"]
zapatos = ["Z1", "Z2"]


# Con product Generamos todas las combinaciones posibles de outfit
# es el producto cartesiano

# product devuelve un iterador, no una lista, por eso usamos list para convertir elresultado en una lista
# ahora autfits queda como una lista de tuplas
outfits = list(product(camisetas, pantalones, zapatos))

# Mostramos el resultado
for outfit in outfits:
    print(outfit)


# vemos el numero total de combinaciones
# es como si multiplicaramos camisetas x pantalones x zapatos
print("Número total de outfits:", len(outfits))
```

### Generación de combinaciones

1. Genera todas las combinaciones posibles de outfits (camiseta, pantalón, zapatos).
2. Representa el resultado en forma de lista o tabla en Python y en R.
3. Verifica que el número total de combinaciones coincide con el resultado obtenido en el principio multiplicativo.

```python

# con append se puede añadir una camiseta nueva
camisetas.append("C4")

# Recalculamos combinaciones
# se tienen que volver a calcuar con los nuevos añadidos
outfits_actualizados = list(product(camisetas, pantalones, zapatos))

print("Nuevo número de outfits:", len(outfits_actualizados))

# Añadimos un nuevo par de zapatos
# lo hacemos igual que antes cuando añadimos una camiseta
zapatos.append("Z3")

# Recalculamos combinaciones
outfits_final = list(product(camisetas, pantalones, zapatos))

print("Número final de outfits:", len(outfits_final))
```

```python

# representamos los outfits en una tabla de pandas convirtiendo los datos en un dataframe
df = pd.DataFrame(outfits_final, columns=["Camiseta", "Pantalón", "Zapatos"])

print(df.head())
```
